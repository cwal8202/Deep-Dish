```python
# 특정 브랜드 & 메뉴 유지하고 나머지 삭제하기 함수 설정
def keep_only_selected_menus(df,brand_menu_dict):
    """
    brand_menu_dict에 지정된 브랜드와 메뉴만 유지하고 나머지는 삭제합니다.
    
    Parameters:
        df (pd.DataFrame): 원본 데이터프레임
        brand_menu_dict (dict): {브랜드명: [남길 메뉴명 리스트]} 구조
        
    Returns:
        pd.DataFrame: 필터링된 결과
    """
    # 브랜드와 메뉴 조건 필터링
    conditions = []
    for brand,menu_list in brand_menu_dict.items():
        condition = (df['brand'] == brand) & (df['name'].isin(menu_list))
        conditions.append(condition)
    
    # 위의 조건들 OR로 결합하여 대조건 생성 = 지정된 브랜드의 지정된 메뉴만 남음
    conbined_condition = conditions[0]
    for cond in conditions[1:]:
        conbined_condition |= cond
    return df[conbined_condition].reset_index(drop=True)
```

```python
# 환경 변수 체크
from dotenv import load_dotenv
import os

load_dotenv()
```

```python
import os
import time
import torch
import torch.nn as nn
from PIL import Image
from urllib.request import urlretrieve
from torchvision import transforms, models
import requests

# [1] 네이버 API 정보
client_id = os.getenv('Client_ID')
client_secret = os.getenv('Client_Secret')

# [2] 설정 ★★★★★★★★★★★여기 바꾸고
base_dir = r'C:\Users\Admin\Desktop\햄버거 브랜드 이미지'
max_images_per_menu = 1000
exclude_keywords_in_title = ["쇼핑몰", "츄리링", "바지", "의상", "하의", "상의", "포스터"]

# [3] 검색어 로드
with open('search_texts_last.txt', 'r', encoding='utf-8') as f:
    raw_line = f.read().strip()
    search_keywords = [kw.strip() for kw in raw_line.split(',') if kw.strip()]

# [4] 이진 분류 모델 로딩
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model = models.efficientnet_b0(weights='DEFAULT')
model.classifier = nn.Sequential(
    nn.Dropout(0.5),
    nn.Linear(1280, 1)
)
model.load_state_dict(torch.load("efficientNetB0_0712_test3_best_burger_model.pth", map_location=device))
model = model.to(device)
model.eval()

transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.485, 0.456, 0.406],
                         std=[0.229, 0.224, 0.225])
])

# [5] 햄버거 이미지 판별 함수
def is_hamburger(image_path, threshold=0.5):
    image = Image.open(image_path).convert("RGB")
    image = transform(image).unsqueeze(0).to(device)
    with torch.no_grad():
        output = model(image)
        prob = torch.sigmoid(output).item()
    return prob > threshold

# [6] 안전한 폴더명 생성
def safe_folder_name(name):
    unsafe_chars = ['<', '>', ':', '"', '/', '\\', '|', '?', '*']
    for char in unsafe_chars:
        name = name.replace(char, '_')
    return name.strip()

# [7] 폴더 생성 함수
def create_folder_if_not_exists(folder_path):
    if not os.path.exists(folder_path):
        os.makedirs(folder_path)
        print(f"📁 폴더 생성: {folder_path}")
        return True
    return False

# [8] 네이버 이미지 검색
def search_img_naver(query, display=100, start=1):
    url = "https://openapi.naver.com/v1/search/image"
    headers = {
        "X-Naver-Client-Id": client_id,
        "X-Naver-Client-Secret": client_secret
    }
    params = {
        "query": query,
        "display": display,
        "start": start,
        "sort": "sim",
        "filter": "large"
    }

    try:
        response = requests.get(url, headers=headers, params=params)
        if response.status_code == 200:
            items = response.json().get('items', [])
            return [(item['link'], item['title']) for item in items]
        else:
            print(f"[{query}] 요청 실패: {response.status_code}")
            return []
    except Exception as e:
        print(f"[{query}] API 호출 에러: {e}")
        return []

# [9] 이미지 저장 + 햄버거 판별 필터
def save_image_with_filter(url, save_path, count, temp_dir="temp_img"):
    try:
        os.makedirs(temp_dir, exist_ok=True)
        ext = url.split('.')[-1].split('?')[0].lower()
        if ext not in ['jpg', 'jpeg', 'png', 'gif', 'webp']:
            ext = 'jpg'
        temp_path = os.path.join(temp_dir, f"temp_img.{ext}")
        urlretrieve(url, temp_path)

        if is_hamburger(temp_path):
            filename = f"naver_img_{count:04d}.{ext}"
            final_path = os.path.join(save_path, filename)
            os.rename(temp_path, final_path)
            return True
        else:
            print("❌ 비햄버거 이미지로 분류됨 → 저장 안 함")
            os.remove(temp_path)
            return False

    except Exception as e:
        print(f"→ 저장 실패: {e}")
        return False

# [10] 실행
print("=" * 60)
print(f"🚀 이미지 크롤링 시작 - 메뉴당 최대 {max_images_per_menu}장")
print("=" * 60)

total_success = 0
total_failed = 0

for idx, keyword in enumerate(search_keywords, 1):
    print(f"\n[{idx}/{len(search_keywords)}] 🔍 검색어: '{keyword}'")

    try:
        parts = keyword.split(' ', 1)
        if len(parts) < 2:
            print(f"❌ 잘못된 포맷: {keyword}")
            total_failed += 1
            continue

        brand, menu = parts[0], parts[1]
        safe_brand = safe_folder_name(brand)
        safe_menu = safe_folder_name(menu)
        target_path = os.path.join(base_dir, safe_brand, safe_menu)
        create_folder_if_not_exists(target_path)

        saved_count = 0
        start = 1

        while saved_count < max_images_per_menu:
            display_count = min(100, max_images_per_menu - saved_count)
            image_items = search_img_naver(keyword, display=display_count, start=start)

            if not image_items:
                print("💧 더 이상 검색 결과 없음")
                break

            for link, title in image_items:
                if any(ex_kw in title.lower() for ex_kw in exclude_keywords_in_title):
                    print(f"→ 제외됨 (필터): {title[:50]}...")
                    continue
                if save_image_with_filter(link, target_path, saved_count + 1):
                    saved_count += 1
                    print(f"✅ 저장 성공: {saved_count}/{max_images_per_menu}")
                if saved_count >= max_images_per_menu:
                    break
                time.sleep(0.1)

            start += 100
            if len(image_items) < 100:
                break

        print(f"📊 [{keyword}] 저장 완료: {saved_count}장")
        total_success += saved_count
        time.sleep(1)

    except Exception as e:
        print(f"❌ [{keyword}] 처리 중 오류 발생: {e}")
        total_failed += 1

# [11] 로그 저장
print("\n" + "=" * 60)
print("🎉 크롤링 완료!")
print(f"📊 총 저장된 이미지: {total_success}장")
print(f"❌ 실패한 검색어: {total_failed}개")
print("=" * 60)

log_filename = os.path.join(base_dir, 'crawling_log.txt')
with open(log_filename, 'w', encoding='utf-8') as f:
    f.write(f"네이버 이미지 크롤링 결과\n")
    f.write(f"처리 시간: {time.strftime('%Y-%m-%d %H:%M:%S')}\n")
    f.write(f"총 검색어 수: {len(search_keywords)}\n")
    f.write(f"총 저장 이미지: {total_success}장\n")
    f.write(f"실패한 검색어: {total_failed}개\n")
    f.write("=" * 50 + "\n")

print(f"📝 로그 저장 완료: {log_filename}")
```
